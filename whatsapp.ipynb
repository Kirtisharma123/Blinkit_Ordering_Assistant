```python
from blinkit_automation.driver_utils import init_driver
import time
from selenium.webdriver.common.by import By

driver = init_driver()
driver.get('https://web.whatsapp.com')
time.sleep(5)
```

# Code to send a message

```python
message='test'

btn_grocery_bot = driver.find_element(by=By.XPATH, value='//*[@id="pane-side"]/div/div/div/div[1]/div/div/div/div[2]/div[1]/div[1]/span')
btn_grocery_bot.click()

def find(xpath):
    return driver.find_element(by=By.XPATH, value=xpath)

ip_send_msg = find('//*[@id="main"]/footer/div[1]/div/span/div/div[2]/div/div[3]/div[1]')
ip_send_msg.click()
ip_send_msg.send_keys(message)
btn_send = find('//*[@id="main"]/footer/div[1]/div/span/div/div[2]/div/div[4]/button/span')
btn_send.click()
```
